```python
import matplotlib.pyplot as plt
```

#### Lineplot
1. Trends over time
2. Time series analysis
3. Upward/downword trend

```python
import numpy as np
```

```python
overs = np.arange(1, 21)
MI = np.random.randint(0, 37, 20)
RCB = np.random.randint(0, 37, 20)
```

```python
MI
```

```python
MI_Score = np.cumsum(MI)
MI_Score
```

```python
RCB_Score = np.cumsum(RCB)
RCB_Score
```

```python
plt.figure(figsize=(7,3))
plt.plot(overs, MI_Score, marker = 'o', color= 'blue', label='MI')
plt.plot(overs, RCB_Score, marker= '<', color ='red', label='RCB')
plt.xlabel('Overs')
plt.ylabel('Score')
plt.title('Match Summary')
plt.legend()
plt.show()
```

#### Scatterplot
1. Relationship between two variables
2. Correlation
3. Positive correlation
4. Negative correlation

```python
plt.figure(figsize=(7,4))
plt.scatter(overs, MI, color= 'blue', marker='o', label='MI')
plt.scatter(overs, RCB, color='Red', marker='<', label='RCB')
plt.xlabel('Overs')
plt.ylabel('Score')
plt.title('Match Summary')
plt.legend()
plt.show()
```

#### Pie chart
1. showing percentage distribution
2. part to whole comparision
3. best to use when categories <= 5
   

```python
marks = [70, 90, 67, 40, 76, 72]
subjects = ['m1', 'm2', 'm3', 'm4', 'm5', 'm6']
```

```python
plt.figure(figsize=(7,4))
plt.pie(marks, labels = subjects, autopct = '%0.2f%%')
plt.show()
```
