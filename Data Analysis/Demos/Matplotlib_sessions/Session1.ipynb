```python
import matplotlib.pyplot as plt
```

#### Bar plot
1. Compairing categories
2. Showing frequencies or count

```python
products = ['Hot coffee', 'Cold coffee', 'Tea', 'Salad']
summer = [30, 90, 80, 50]
```

```python
plt.figure(figsize= (7, 3))
plt.bar(products, summer, color = 'grey')
plt.title('Sales of product for summer')
plt.xlabel('Products')
plt.ylabel('Count')
plt.show()
```

#### Histogram 
1. Distribution of numerical values
2. Frequencies of value ranges
3. To check outliers with 3 sigma rule

```python
import numpy as np
```

```python
age = np.random.randint(1, 120, 1000)
```

```python
plt.figure(figsize= (7, 3))   #(width, height)
plt.hist(age, color = 'red')
plt.show()
```
