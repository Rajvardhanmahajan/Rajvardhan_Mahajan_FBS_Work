```python
import sklearn
print(sklearn.__version__)
```

```python
import numpy as np
from sklearn.preprocessing import OrdinalEncoder
```

```python
data = np.array([['Low'], ['Medium'], ['High'], ['Low'], ['Medium']])
data
```

```python
OE = OrdinalEncoder(categories = [['Low','Medium','High']], dtype = int)
```

```python
OE.fit_transform(data)
```

#### One-Hot Encoding

##### OneHotEncoder Class

```python
data = np.array([['IT'], ['HR'], ['Sales']])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
OHE = OneHotEncoder(sparse_output=False, dtype = int)
```

```python
OHE.fit_transform(data)
```
