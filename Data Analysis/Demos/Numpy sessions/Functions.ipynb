```python
arr1 = np.arange(1, 10).reshape((3,3))
arr2 = np.arange(3, 12).reshape((3,3))
arr1, arr2
```

```python
np.add(arr2, arr1)
```

```python
arr2 + arr1
```

```python
np.subtract(arr2, arr1)
```

```python
arr2 - arr1
```

```python
np.multiply(arr2, arr1)
```

```python
arr2 * arr1
```

```python
np.divide(arr2, arr1)
```

```python
arr2 / arr1
```
