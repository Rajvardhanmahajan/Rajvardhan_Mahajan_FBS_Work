```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Month':['Jan', 'Feb', 'March'],
    'Sales':[100, 150, 140]}
```

```python
df = pd.DataFrame(data)
plt.plot(df['Month'], df['Sales'])
plt.show()
```

#### pandas

```python
dataa = {'Department' : ['IT', 'HR', 'IT', 'HR'],
        'Employee' : ['Rajvardhan', 'Harshwardhan', 'Shubham', 'kunal'],
        'Salary' : [50000, 70000, 40000, 60000]}
df = pd.DataFrame(dataa)
```

```python
df.groupby('Department')['Salary'].mean()
```

```python
import plotly.express as px

months = ['Jan', 'Feb', 'Mar', 'Apr']
sales = [10000, 15000, 12000, 20000]

fig = px.bar(
    x=months,
    y=sales,
    title='Monthly Sales Analysis'
)

fig.show()
```
